```python
from wavesongs.utils import plot
from wavesongs.object import Synthetic
from wavesongs.data.manager import ProjDirs

proj_dirs = ProjDirs(audios="./assets/audios", results="./experiments/")
ploter = plot.set_plotter(library="matplotlib", save_dir=str(proj_dirs.images))

import numpy as np
sr = 44100
t = np.linspace(0, 1, sr)
x = np.sin(2 * np.pi * 2000 * t) # + np.sin(2 * np.pi * 3500 * t)

synthetic = Synthetic(duration=1, proj_dirs=proj_dirs, sr=sr)
synthetic.initialize(x)

synthetic.acoustical_features(umbral_FF=1.4, n_fft=512, ff_method="yin")
_ = ploter.spectrogram(synthetic)#, ylim=(0, 0.5))
```

```python
from wavesongs.core.bird import Model

duration = 2*np.pi
model = Model(ovsr=1)
t = np.linspace(0, duration, int(duration*sr))

synthetic = Synthetic(duration=duration, proj_dirs=proj_dirs, sr=sr)
synthetic.initialize()

ploter._COLORS["motor_gesture"] = "Blues"
```

```python
np.linspace(0, 1.75, 8)
```

```python
for i in np.linspace(0, 1.75, 2):# range(0, 360, 45):
     z = {"a0": 0.03, "a1":0.02, "a2": 1, "a3":0,
          "b0": 0.8, "b1": 0.2, "b2": 1, "b3": i}
     synthetic.z = z

     alpha = z["a0"] + z["a1"] * np.cos((z["a2"] * t + np.pi*z["a3"])) # a0 + a1*sin(2pi a2 t)
     beta  = z["b0"] + z["b1"] * np.cos((z["b2"] * t + np.pi*z["b3"])) # b0 + b1*sin(2pi (b2*t+ b3))
     curves = [alpha, beta]

     synthetized = model.motor_gesture(synthetic, curves)

     print(f"Angle = {i} rad / {round(i*180/np.pi, 2)}°\n"+
           f"alpha = {z['a0']} + {z['a1']} * cos({z['a2']} * t + {z['a3']})\n"
           f"beta = {z['b0']} + {z['b1']} * cos({z['b2']} * t + {z['b3']})")
     # Plot
     save_name = f"Hphi{i}-"
     _ = ploter.alpha_beta(synthetized, save=True, save_name=save_name, ylim=(0, 1))

     # Plot waveforme generated
     synthetized.acoustical_features(umbral_FF=1.4, n_fft=512, ff_method="yin")
     _ = ploter.spectrogram(synthetized, save=True, save_name=save_name) #, ylim=(0, 0.5))
     synthetized.play()
```
